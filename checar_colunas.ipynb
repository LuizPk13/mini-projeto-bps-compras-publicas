# Checagem de colunas — bases BPS 2020-2026

```python
# Biblioteca para leitura e manipulação dos dados.
import pandas as pd
# Biblioteca para montar o caminho dos arquivos.
import os

# Pasta onde estão os 7 arquivos CSV baixados.
pasta = "bases_20_26"
# Lista dos anos a conferir.
anos = [2020, 2021, 2022, 2023, 2024, 2025, 2026]

# Dicionário para guardar as colunas encontradas.
colunas_por_ano = {}
```

```python
# Passando por cada ano, um de cada vez.
for ano in anos:
    # Montando o caminho completo do arquivo daquele ano.
    caminho = os.path.join(pasta, f"{ano}.csv")
    try:
        # Lendo só as 5 primeiras linhas, com separador ";".
        try:
            df = pd.read_csv(caminho, sep=";", nrows=5, encoding="utf-8", low_memory=False)
        # Se não der certo com ";", tenta com "," (caso o arquivo venha em outro formato).
        except Exception:
            df = pd.read_csv(caminho, sep=",", nrows=5, encoding="utf-8", low_memory=False)

        # Guardando a lista de colunas desse ano no dicionário.
        colunas_por_ano[ano] = list(df.columns)
        # Mostrando na tela quantas colunas esse ano tem e quais são.
        print(f"\n=== {ano} ===")
        print(f"Nº de colunas: {len(df.columns)}")
        print(list(df.columns))
    # Se der qualquer erro na leitura, mostra a mensagem em vez de travar o notebook.
    except Exception as e:
        print(f"\n=== {ano} ===")
        print(f"ERRO ao ler: {e}")
```

```python
# Título da seção de comparação entre os anos.
print("\n\n=== COMPARAÇÃO ===")

# Conjunto (sem repetição) com todas as colunas que apareceram em qualquer ano.
todas_colunas = set()
for cols in colunas_por_ano.values():
    todas_colunas.update(cols)

# Mostrando quantas colunas diferentes existem no total, somando todos os anos.
print(f"Total de colunas distintas encontradas em todos os anos: {len(todas_colunas)}\n")
```

```python
# Para cada coluna encontrada, em ordem alfabética...
for col in sorted(todas_colunas):
    # Lista dos anos em que essa coluna aparece.
    presente_em = [ano for ano, cols in colunas_por_ano.items() if col in cols]
    # Lista dos anos em que essa coluna NÃO aparece.
    ausente_em = [ano for ano in anos if ano not in presente_em]
    # Só mostra na tela se a coluna faltar em pelo menos um ano
    # (colunas presentes em todos os anos não precisam de aviso).
    if ausente_em:
        print(f"'{col}' -> presente em {presente_em} | AUSENTE em {ausente_em}")
```
