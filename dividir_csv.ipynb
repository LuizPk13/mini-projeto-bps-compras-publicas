# Dividindo um CSV em duas partes para fazer um upload para o BigQuery

```python
import pandas as pd

df = pd.read_csv(
    "BPS_20_26_LuizFernandoDeJesusSilvaHomem.csv",
    sep=";",
    encoding="utf-8",
    low_memory=False,
    dtype=str  # lê TUDO como texto, sem reformatar nada
)

metade = len(df) // 2
parte1 = df.iloc[:metade]
parte2 = df.iloc[metade:]

parte1.to_csv("BPS_parte1.csv", index=False, sep=";", encoding="utf-8")
parte2.to_csv("BPS_parte2.csv", index=False, sep=";", encoding="utf-8")

print(f"Parte 1: {len(parte1)} linhas")
print(f"Parte 2: {len(parte2)} linhas")
```
